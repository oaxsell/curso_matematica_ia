# Usar o matplotlib com o retorno financeiro

```python
# Aplicando a formula:
# O cálculo é feito de forma vetorizada no NumPy.
# A série de valores de preço da ação foi convertida em um vetor.
# Para aplicar a fórmula em todos os elementos de uma vez, 
# utilizamos a técnica de fatiamento (slicing), 
# que cria versões deslocadas do vetor original.
# Portanto, vetor[1:] = valor final vetor[:-1] = valor inicial
ret = (vetor[1:] - vetor[:-1]) / vetor[:-1]
```

```python
import matplotlib.pyplot as plt
```

```python
indices=np.arange(len(ret))
plt.plot(
    indices, 
    ret, 
    label='Variação do Preço da Ação', 
    color='black',
    marker='o',
    linewidth=2,
    linestyle='-')
plt.title("Variação do Preço da Ação vs. Índice do Array")
plt.xlabel("Índice")
plt.ylabel("Variação do Preço da Ação")
plt.xticks(indices)
plt.legend()
plt.grid(True, linestyle='-', alpha=0.6)
plt.show()
```
